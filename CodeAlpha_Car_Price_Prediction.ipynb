```python
# Car Price Prediction

from sklearn.linear_model import LinearRegression
import numpy as np

# Sample data
X = np.array([[10000,5],[20000,3],[30000,2],[40000,1]])
y = np.array([3,5,7,9])

# Train model
model = LinearRegression()
model.fit(X,y)

# Predict
print("Predicted Price:", model.predict([[25000,3]]))
```
